## 📦 Setup

```python
# Add parent directory to path
import sys
sys.path.insert(0, '..')

# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Import our modules
from src.data.loaders import load_wyscout_data
from src.data.extractors import extract_set_pieces, calculate_success_metrics
from src.visualization.pitch import draw_pitch, plot_heatmap, plot_set_piece

print("✅ All imports successful!")
```

## 1️⃣ Load Data

```python
# Load World Cup 2022 data
data = load_wyscout_data()

print(f"\n📊 Dataset Info:")
print(f"   Shape: {data.shape}")
print(f"   Columns: {len(data.columns)}")
print(f"   Matches: {data['match_id'].nunique()}")
```

## 2️⃣ Extract Set Pieces

```python
# Extract set pieces from the data
set_pieces = extract_set_pieces(data)

print(f"\n⚽ Found {len(set_pieces)} set pieces!")
print(f"\n📊 Distribution by Type:")
print(set_pieces['type'].value_counts())
```

```python
# Calculate success metrics
metrics = calculate_success_metrics(set_pieces)

print("\n🎯 Success Metrics:")
print(f"   Goal Rate: {metrics['goal_rate']:.2f}%")
print(f"   Shot Rate: {metrics['shot_rate']:.2f}%")
print(f"   Success Rate: {metrics['success_rate']:.2f}%")
print(f"   Possession Retained: {metrics['possession_retained_rate']:.2f}%")
```

## 3️⃣ Visualize

```python
# Draw a clean pitch
fig, ax = draw_pitch()
ax.set_title('FIFA Standard Football Pitch', fontsize=14, fontweight='bold')
plt.show()
```

```python
# Create a heatmap of set piece locations
fig, ax = plot_heatmap(set_pieces, zone_type='receiver', title='Set Piece Locations')
plt.show()
```

```python
# Visualize a single set piece
sample_sp = set_pieces.iloc[0].to_dict()
fig, ax = plot_set_piece(sample_sp)
plt.show()
```

## 4️⃣ Simple Prediction

```python
# Import feature engineering and model
from src.features.spatial import calculate_spatial_features
from src.models.receiver_predictor import FirstReceiverPredictor

# Add features
set_pieces_feat = calculate_spatial_features(set_pieces)

# Create and train a simple model
predictor = FirstReceiverPredictor(n_estimators=50, max_depth=4)
X, _ = predictor.prepare_features(set_pieces_feat)
y = set_pieces_feat['first_receiver_idx'].fillna(0).astype(int)

# Train
metrics = predictor.train(X, y)

print(f"\n🎯 Model trained successfully!")
print(f"   Validation Accuracy: {metrics['val_accuracy']:.2%}")
```

```python
# Show feature importance
importance = predictor.get_feature_importance()

# Plot top 10 features
plt.figure(figsize=(10, 6))
top_10 = importance.head(10)
plt.barh(top_10['feature'], top_10['importance'], color='steelblue')
plt.xlabel('Importance')
plt.title('Top 10 Most Important Features', fontweight='bold')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```
